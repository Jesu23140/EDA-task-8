```python
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
  df[col] = (
      df[col]
      .astype('string')
      .str.strip()
  )

for col in categorical_cols:
  print(f"\n{col}:")
  print(df[col].unique())
```

```python
score_cols = [
    'math score',
    'reading score',
    'writing score',
]

stats = df[score_cols].agg(
    ['mean', 'median', 'std']
).T

stats['Q1'] = df[score_cols].quantile(0.25)
stats['Q2'] = df[score_cols].quantile(0.50)
stats['Q3'] = df[score_cols].quantile(0.75)

print(stats.round(2))
```

```python
df['total marks'] = df[['math score', 'reading score', 'writing score']].sum(axis=1)
df['percentage'] = df['total marks'] / 300 * 100
print(df[['math score', 'reading score', 'writing score', 'total marks', 'percentage']].head())
print(df[['total marks', 'percentage']].describe())
```

```python
subjects = ['math score', 'reading score', 'writing score']
print(df[subjects].describe())
print("Median:")
print(df[subjects].median())
```

```python
for col in subjects:
    q1, q3 = df[col].quantile([.25, .75])
    outliers = df[(df[col] < q1 - 1.5 * (q3 - q1)) | (df[col] > q3 + 1.5 * (q3 - q1))]
    print(col, "Outliers:", len(outliers))
df.to_csv("Student_Performance_Cleaned.csv", index=False)
print("Task completed!")
```

```python
education_avg = df.groupby('parental level of education')[subjects].mean()
education_avg.plot(kind='bar', figsize=(10, 5))
plt.title('Average Scores by Parental Education'); plt.xlabel('Parental Education Level'); plt.ylabel('Average Score')
plt.xticks(rotation=45, ha='right'); plt.legend(title='Subject'); plt.tight_layout(); plt.show()
```

```python
lunch_avg = df.groupby('lunch')[subjects].mean()
lunch_avg.plot(kind='bar', figsize=(8, 5))
plt.title('Average Scores by Lunch Type'); plt.xlabel('Lunch Type'); plt.ylabel('Average Score')
plt.xticks(rotation=0); plt.legend(title='Subject'); plt.tight_layout(); plt.show()
```

```python
for subject in subjects:
    plt.figure(figsize=(7, 5))
    sns.boxplot(data=df, x='test preparation course', y=subject)
    plt.title(f'{subject.title()} by Test Preparation'); plt.xlabel('Test Preparation Course'); plt.ylabel('Score')
    plt.tight_layout(); plt.show()
```

```python
for x, y, title in [('math score', 'reading score', 'Math vs Reading Scores'), ('math score', 'writing score', 'Math vs Writing Scores'), ('reading score', 'writing score', 'Reading vs Writing Scores')]:
    sns.scatterplot(data=df, x=x, y=y)
    plt.title(title); plt.show()
```

```python
correlation = df[subjects].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Between Subject Scores'); plt.tight_layout(); plt.show()
```
